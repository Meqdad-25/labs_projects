```python
# Imports
import pandas as pd
```

```python
# Load training data
train = pd.read_csv('./data/data.csv')

# Drop rows with no rent
train = train.dropna(subset=['rent'])

# Our single guess for everyone: the median rent
median_rent = train['rent'].median()
print('Median rent:', median_rent)
```

```python
train.head()
```

```python
# Load the test file
test = pd.read_csv('test.csv')

# Predict the same median rent for every row
submission = pd.DataFrame({
    'Property_id': test['Property_id'],
    'rent': median_rent
})

# Save in the required format
submission.to_csv('submission.csv', index=False)
submission.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

# Split the training data
train_part, val_part = train_test_split(train, test_size=0.2, random_state=42)

# Guess = median rent of the training part
guess = train_part['rent'].median()

# How far off is that guess on the validation part?
val_mae = mean_absolute_error(val_part['rent'], [guess] * len(val_part))
print('Local MAE:', round(val_mae, 2), 'BHD')
```
